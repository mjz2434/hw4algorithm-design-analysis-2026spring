## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:
算法思想：动态规划 / Kadane 算法

设：cur表示以当前第 i 天作为卖出日时，能够获得的最大连续收益。

遍历数组时，有两种选择：

情况 1：继续持有之前买入的股票 cur=cur+profits[i]

情况 2：放弃之前的区间，从当前附近重新买入

如果之前累计收益已经拖累结果，就重新开始。因为题目要求买入日和卖出日不同，所以至少要包含两天收益。

因此可以维护：cur=max(cur+profits[i], profits[i−1]+profits[i])

其中：cur+profits[i]：继续原来的买入日期

profits[i−1]+profits[i]：从第 i−1 天重新买入，第 i 天卖出

算法时间复杂度：O(n)

````python
# add your code here
# your algorithm time complexity is: O(n)
def max_stock_profit(profits):
    n = len(profits)

    if n < 2:
        return None

    cur_sum = profits[0] + profits[1]
    cur_start = 0

    max_sum = cur_sum
    buy_day = 0
    sell_day = 1

    for i in range(2, n):
        extend_sum = cur_sum + profits[i]

        restart_sum = profits[i - 1] + profits[i]

        if restart_sum > extend_sum:
            cur_sum = restart_sum
            cur_start = i - 1
        else:
            cur_sum = extend_sum

        if cur_sum > max_sum:
            max_sum = cur_sum
            buy_day = cur_start
            sell_day = i
        
    return max_sum, buy_day + 1, sell_day + 1

profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
ans = max_stock_profit(profits)

print("最大收益:", ans[0])
print("买入日期: 第", ans[1], "天")
print("卖出日期: 第", ans[2], "天")
````

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:1.算法思想：二分查找

对于中间位置 mid，判断它和左右邻居的关系。

如果： A[mid] < A[mid−1] 且 A[mid] < A[mid+1] ,那么 A[mid] 就是局部最小值。

否则分两种情况：

情况 1：左边更小

如果： A[mid] > A[mid−1] ,说明从 mid 往左是下降方向，因此左侧一定存在局部最小值。于是：right=mid−1

情况 2：右边更小

如果： A[mid] > A[mid+1] ,说明从 mid 往右是下降方向，因此右侧一定存在局部最小值。于是：left=mid+1

````python
# add your code here
# your algorithm time complexity is: O(logn)
def find_one_local_min(A):
    n = len(A)

    if n < 3:
        return None

    left = 1
    right = n - 2

    while left <= right:
        mid = (left + right) // 2

        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid], mid

        elif A[mid] > A[mid - 1]:
            right = mid - 1

        else:
            left = mid + 1

    return None

A = [9, 3, 7, 2, 1, 4, 5]

value, index = find_one_local_min(A)

print("局部最小值:", value)
print("下标:", index)
````

idea:2.算法思想

如果要找出所有局部最小值，就不能只用二分，因为局部最小值可能有很多个。

直接遍历数组中间的每一个元素： 2≤x≤n−1

判断是否满足： A[x]<A[x−1]

且： A[x]<A[x+1] ,如果满足，就加入答案。

````python
# add your code here
# your algorithm time complexity is: O(n)
def find_all_local_min(A):
    n = len(A)

    result = []

    for i in range(1, n - 1):
        if A[i] < A[i - 1] and A[i] < A[i + 1]:
            result.append((A[i], i))

    return result

A = [9, 3, 7, 2, 1, 4, 5]

ans = find_all_local_min(A)

print("所有局部最小值:")
for value, index in ans:
    print("值:", value, "下标:", index)
````

3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:1.直接把整个数组排序，然后取排序后的前 k 个元素。

````python
# add your code here
# your algorithm time complexity is: O(nlogn)
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


def k_smallest_by_sort(arr, k):
    sorted_arr = merge_sort(arr)
    return sorted_arr[:k]


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

print(k_smallest_by_sort(arr, k))
````

idea:2.建立一个包含所有 n 个元素的最小堆。最小堆的性质是：堆顶元素一定是当前堆中的最小值。因此，只需要连续执行 k 次 extract_min，就可以按升序取出最小的 k 个数。

````python
# add your code here
# your algorithm time complexity is: O(nlogn)
class MinHeap:
    def __init__(self):
        self.heap = []

    def insert(self, x):
        self.heap.append(x)
        self._sift_up(len(self.heap) - 1)

    def extract_min(self):
        if not self.heap:
            return None

        min_value = self.heap[0]
        last_value = self.heap.pop()

        if self.heap:
            self.heap[0] = last_value
            self._sift_down(0)

        return min_value

    def _sift_up(self, index):
        while index > 0:
            parent = (index - 1) // 2

            if self.heap[parent] <= self.heap[index]:
                break

            self.heap[parent], self.heap[index] = self.heap[index], self.heap[parent]
            index = parent

    def _sift_down(self, index):
        n = len(self.heap)

        while True:
            left = 2 * index + 1
            right = 2 * index + 2
            smallest = index

            if left < n and self.heap[left] < self.heap[smallest]:
                smallest = left

            if right < n and self.heap[right] < self.heap[smallest]:
                smallest = right

            if smallest == index:
                break

            self.heap[index], self.heap[smallest] = self.heap[smallest], self.heap[index]
            index = smallest


def k_smallest_by_min_heap(arr, k):
    heap = MinHeap()

    for x in arr:
        heap.insert(x)

    result = []

    for _ in range(k):
        result.append(heap.extract_min())

    return result


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

print(k_smallest_by_min_heap(arr, k))
````

idea:3.如果数组非常大，无法全部放入内存，但计算机可以存储 k 个数据，那么应该维护一个大小为 k 的最大堆。最大堆堆顶保存的是当前这 k 个数里的最大值。

遍历数据流时：
前 k 个数直接放入最大堆；
从第 k+1 个数开始，假设当前数为 x；
如果 x 大于等于堆顶，说明它不可能进入最小的 k 个数，直接丢弃；
如果 x 小于堆顶，说明它比当前最大的候选数更小，用它替换堆顶；
最后堆中剩下的 k 个数就是整个数据中的最小 k 个数；
最后把这 k 个数排序输出即可。

````python
# add your code here
# your algorithm time complexity is: O(nlogk+klogk)
import heapq


def k_smallest_big_data(data_stream, k):
    if k <= 0:
        return []

    max_heap = []

    for x in data_stream:
        if len(max_heap) < k:
            heapq.heappush(max_heap, -x)

        else:
            current_max = -max_heap[0]

            if x < current_max:
                heapq.heapreplace(max_heap, -x)

    result = [-x for x in max_heap]

    result.sort()

    return result


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

print(k_smallest_big_data(arr, k))
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:用栈来模拟星球的碰撞过程。因为星球按数组顺序排列，只有当左边星球向右运动、右边星球向左运动时才会发生碰撞，也就是栈顶元素为正数、当前元素为负数的情况。

遍历数组中的每个星球，将已经确定不会消失的星球放入栈中。遇到一个向左运动的星球时，就不断和栈顶向右运动的星球比较质量。如果栈顶星球质量较小，栈顶星球被撞毁，继续比较下一个栈顶；如果栈顶星球质量较大，当前星球消失；如果两者质量相同，根据题意，向左运动的当前星球消失，栈顶向右运动的星球保留。遍历结束后，栈中剩下的元素就是最终仍然存在的星球状态。

````python
# add your code here
# your algorithm time complexity is: O(n)
def planet_collision(A):
    stack = []

    for x in A:
        alive = True

        while alive and stack and stack[-1] > 0 and x < 0:
            top = stack[-1]

            if abs(top) < abs(x):
                stack.pop()

            elif abs(top) > abs(x):
                alive = False

            else:
                alive = False

        if alive:
            stack.append(x)

    return stack


A = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
print(planet_collision(A))

A = [23, -8, 9, -3, -7, 9, -23, 22]
print(planet_collision(A))
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:使用动态规划。

定义：dp[i][w] 表示：只考虑前 i 块大理石，在承重不超过 w 的情况下，能够获得的最大价值。

对于第 i 块大理石，有两种选择：

第一种是不装它，那么价值为：dp[i−1][w]

第二种是装它，前提是当前容量 w 能装下它。设它的重量为 weight，价值为 value，则价值为：dp[i−1][w−weight]+value

所以状态转移方程为：dp[i][w]=max(dp[i−1][w], dp[i−1][w−weight]+value)

最终答案就是：dp[n][capacity]

````python
# add your code here
# your algorithm time complexity is: O(nW)
def knapsack_01(stones, capacity):
    n = len(stones)

    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight, value = stones[i - 1]

        for w in range(capacity + 1):
            dp[i][w] = dp[i - 1][w]

            if w >= weight:
                dp[i][w] = max(
                    dp[i][w],
                    dp[i - 1][w - weight] + value
                )

    selected = []
    w = capacity

    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            weight, value = stones[i - 1]
            selected.append((weight, value))
            w -= weight

    selected.reverse()

    return dp[n][capacity], selected


stones = [(5, 10), (4, 40), (6, 30), (3, 50)]
capacity = 9

max_value, selected_stones = knapsack_01(stones, capacity)

print("最大价值:", max_value)
print("选择的大理石:", selected_stones)
````

输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:利用二叉搜索树的性质来解决。二叉搜索树的中序遍历结果应该是一个递增序列，如果有两个节点的值被错误交换，那么中序遍历时就会出现顺序错误，也就是前一个节点的值大于当前节点的值。

因此，我们对整棵树进行中序遍历，并用一个变量记录当前节点的前一个节点。如果发现 prev.val > cur.val，说明这里出现了逆序。第一次发现逆序时，把 prev 记录为第一个错误节点；之后每次发现逆序，都把 cur 记录为第二个错误节点。遍历结束后，只需要交换这两个节点的值，就可以恢复二叉搜索树，并且不会改变原来的树结构。

````python
# add your code here
# your algorithm time complexity is: O(n)
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def recoverTree(self, root):
        self.prev = None
        self.first = None
        self.second = None

        def inorder(node):
            if not node:
                return

            inorder(node.left)

            if self.prev and self.prev.val > node.val:
                if not self.first:
                    self.first = self.prev
                self.second = node

            self.prev = node

            inorder(node.right)

        inorder(root)

        if self.first and self.second:
            self.first.val, self.second.val = self.second.val, self.first.val


def build_tree(arr):
    if not arr or arr[0] is None:
        return None

    root = TreeNode(arr[0])
    queue = deque([root])
    i = 1

    while queue and i < len(arr):
        node = queue.popleft()

        if i < len(arr) and arr[i] is not None:
            node.left = TreeNode(arr[i])
            queue.append(node.left)
        i += 1

        if i < len(arr) and arr[i] is not None:
            node.right = TreeNode(arr[i])
            queue.append(node.right)
        i += 1

    return root


def tree_to_list(root):
    if not root:
        return []

    result = []
    queue = deque([root])

    while queue:
        node = queue.popleft()

        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)

    while result and result[-1] is None:
        result.pop()

    return result


root = build_tree([1, 3, None, None, 2])
Solution().recoverTree(root)
print(tree_to_list(root))


root = build_tree([3, 1, 4, None, None, 2])
Solution().recoverTree(root)
print(tree_to_list(root))
````

## 问题 9 \*
给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。

请分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:用双指针来做。对于任意一个位置，它能接多少水，取决于它左边最高柱子和右边最高柱子中较矮的那个。也就是说，当前位置的接水量是：min(左边最高柱子,右边最高柱子)−当前柱子高度

如果结果大于 0，就说明这个位置可以接水。为了不额外开数组，可以使用两个指针分别从左右两端向中间移动，同时维护 left_max 和 right_max。每次比较左右两边当前柱子的高度，优先处理较矮的一边，因为较矮的一边的接水上限已经可以确定。处理左边时，如果当前高度低于 left_max，就能接 left_max - height[left] 的水；否则更新 left_max。右边同理。

````python
# add your code here
# your algorithm time complexity is: O(n)
def trap(height):
    n = len(height)

    if n < 3:
        return 0

    left = 0
    right = n - 1

    left_max = 0
    right_max = 0

    ans = 0

    while left < right:
        if height[left] < height[right]:
            if height[left] >= left_max:
                left_max = height[left]
            else:
                ans += left_max - height[left]

            left += 1

        else:
            if height[right] >= right_max:
                right_max = height[right]
            else:
                ans += right_max - height[right]

            right -= 1

    return ans


height = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
print(trap(height))

height = [4, 2, 0, 3, 2, 5]
print(trap(height))
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

 输入：root = [1,2,3], 输出：25

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:使用 深度优先搜索 DFS。

从根节点开始遍历，每向下一层，就把当前路径上的数字更新为：cur=cur×10+node.val

例如当前已经形成数字 12，下一个节点是 3，那么新的数字就是：12×10+3=123

当遍历到叶子节点时，说明一条完整路径已经形成，此时把当前数字加入答案。

````python
# add your code here
# your algorithm time complexity is: O(n)
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build_tree(arr):
    if not arr or arr[0] is None:
        return None

    root = TreeNode(arr[0])
    queue = deque([root])
    i = 1

    while queue and i < len(arr):
        node = queue.popleft()

        if i < len(arr) and arr[i] is not None:
            node.left = TreeNode(arr[i])
            queue.append(node.left)
        i += 1

        if i < len(arr) and arr[i] is not None:
            node.right = TreeNode(arr[i])
            queue.append(node.right)
        i += 1

    return root


def sumNumbers(root):
    def dfs(node, cur):
        if not node:
            return 0

        cur = cur * 10 + node.val

        if not node.left and not node.right:
            return cur

        return dfs(node.left, cur) + dfs(node.right, cur)

    return dfs(root, 0)


root = build_tree([1, 2, 3])
print(sumNumbers(root))

root = build_tree([4, 9, 0, 5, 1])
print(sumNumbers(root))
````
